```python
# go through the visualization bbox_ values 
results = []
for i,row in failed.iterrows():
    im_url = row.url
    parts = Path(im_url).stem.split("_")
    # print(row.url)
    #print(list(Path(f"./images/{parts[0]}/{").iterdir()))
    annot_data = annot[parts[0]]
    # visualizations count for this iamge
    
    for e in annot_data:
        vis_count =0
        im_id = e["image_id"]
        for chart,bboxes in e["visualization_bbox"].items():
            for bbox in bboxes:
                p = Path(f"./images/{parts[0]}/{im_id}.png")
                if p.exists():
                    
                    im = Image.open(f"./images/{parts[0]}/{im_id}.png")
                    subfigure = im.crop(bbox)
                    results.append([parts[0],im_id,vis_count,chart,subfigure])
                    vis_count+=1
                else:
                    print("problem",im_id,parts)
```

```python
uniques = []
mapping = set()
for e in results:
    if f"missing_subs/{e[0]}_{e[1]}_{e[2]}_{e[3]}.png" not in mapping:
        mapping.add(f"missing_subs/{e[0]}_{e[1]}_{e[2]}_{e[3]}.png")
        uniques.append(e)
```
